```python
from oqd_core.interface.analog import (
    AnalogCircuit, AnalogCircuitSSA, AnalogGate, Evolve,
    QuantumDeclaration, ClassicalDeclaration,
    IfElse, RegisterNonZero, While,
    BoolDeclaration, MathExprDeclaration,
    Initialize, Measure, PauliX, PauliZ
)
from oqd_core.interface.analog.register import ClassicalRef
from oqd_core.interface.math import MathNum, MathRef
from oqd_core.compiler.analog.rewrite.lower import lower_analog_circuit
from oqd_compiler_infrastructure import Post, PrettyPrint

printer = Post(PrettyPrint())
```

```python
X = PauliX()
circuit = AnalogCircuit(
    declarations=[QuantumDeclaration(name="q", size=2)],
    sequence=[
        Evolve(gate=AnalogGate(hamiltonian=X * 0.5), duration=2.0),
    ],
)
lowered = lower_analog_circuit(circuit)
print(printer(lowered))
```

```python
X, Z = PauliX(), PauliZ()
circuit = AnalogCircuit(
    declarations=[QuantumDeclaration(name="q", size=2)],
    sequence=[
        Evolve(gate=AnalogGate(hamiltonian=X * 0.5), duration=2.0),
        Evolve(gate=AnalogGate(hamiltonian=Z * 0.3), duration=1.0),
    ],
)
lowered = lower_analog_circuit(circuit)
print(printer(lowered))
```

```python
# IfElse 
X, Z = PauliX(), PauliZ()
circuit = AnalogCircuit(
    declarations=[
        QuantumDeclaration(name="q", size=2),
        ClassicalDeclaration(name="c", size=1),
    ],
    sequence=[
        Evolve(gate=AnalogGate(hamiltonian=X * 0.5), duration=1.0),
        IfElse(
            condition=RegisterNonZero(creg=ClassicalRef(name="c")),
            then_branch=[
                Evolve(gate=AnalogGate(hamiltonian=Z * 0.1), duration=1.0),
            ],
            else_branch=[
                Evolve(gate=AnalogGate(hamiltonian=X * 0.2), duration=1.5),
            ],
        ),
    ],
)
lowered = lower_analog_circuit(circuit)
print(printer(lowered))
```

```python
X = PauliX()
circuit = AnalogCircuit(
    declarations=[
        QuantumDeclaration(name="q", size=2),
        MathExprDeclaration(name="omega", expr=MathNum(value=0.5)),
    ],
    sequence=[
        Evolve(
            gate=AnalogGate(hamiltonian=X * MathRef(name="omega")),
            duration=2.0,
        ),
    ],
)
lowered = lower_analog_circuit(circuit)
print(printer(lowered))
```

```python
X = PauliX()
circuit = AnalogCircuit(
    declarations=[
        QuantumDeclaration(name="q", size=2),
        ClassicalDeclaration(name="c", size=1),
    ],
    sequence=[
        Initialize(),
        Evolve(gate=AnalogGate(hamiltonian=X * 0.5), duration=1.0),
        Measure(),
    ],
)
lowered = lower_analog_circuit(circuit)
print(printer(lowered))
```

```python
X, Z = PauliX(), PauliZ()
circuit = AnalogCircuit(
    declarations=[
        QuantumDeclaration(name="q", size=2),
        ClassicalDeclaration(name="c", size=1),
    ],
    sequence=[
        Evolve(gate=AnalogGate(hamiltonian=X * 0.5), duration=1.0),
        IfElse(
            condition=RegisterNonZero(creg=ClassicalRef(name="c")),
            then_branch=[Evolve(gate=AnalogGate(hamiltonian=Z * 0.1), duration=1.0)],
            else_branch=[Evolve(gate=AnalogGate(hamiltonian=X * 0.2), duration=1.5)],
        ),
        Evolve(gate=AnalogGate(hamiltonian=Z * 0.2), duration=0.5),
    ],
)
lowered = lower_analog_circuit(circuit)
print(printer(lowered))
```

```python
# While
X, Z = PauliX(), PauliZ()
circuit = AnalogCircuit(
    declarations=[
        QuantumDeclaration(name="q", size=2),
        ClassicalDeclaration(name="flag", size=1),
    ],
    sequence=[
        While(
            condition=RegisterNonZero(creg=ClassicalRef(name="flag")),
            body=[
                Evolve(gate=AnalogGate(hamiltonian=Z * 0.3), duration=0.5),
            ],
        ),
    ],
)
lowered = lower_analog_circuit(circuit)
print(printer(lowered))
```

```python
X, Z = PauliX(), PauliZ()
circuit = AnalogCircuit(
    declarations=[
        QuantumDeclaration(name="q", size=2),
        ClassicalDeclaration(name="flag", size=1),
    ],
    sequence=[
        Evolve(gate=AnalogGate(hamiltonian=X * 0.5), duration=1.0),
        While(
            condition=RegisterNonZero(creg=ClassicalRef(name="flag")),
            body=[
                Evolve(gate=AnalogGate(hamiltonian=Z * 0.2), duration=0.5),
            ],
        ),
        Evolve(gate=AnalogGate(hamiltonian=X * 0.1), duration=0.5),
    ],
)
lowered = lower_analog_circuit(circuit)
print(printer(lowered))
```
